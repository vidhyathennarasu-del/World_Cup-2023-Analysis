```python
batting.columns=batting.columns.str.strip().str.lower()
```

```python
bowling.columns = bowling.columns.str.strip().str.lower()
```

```python
matches.columns = matches.columns.str.strip().str.lower()
```

```python
players.columns = players.columns.str.strip().str.lower()
```

```python
batting_players = pd.merge(batting,players,left_on = 'batsman_name',right_on='player_name',how='left')
```

```python
batting_players.head()
```

```python

bowling_players = pd.merge(bowling,players,left_on='bowler_name',right_on = 'player_name',how='left')
bowling_players.head()
```

```python
match_data = pd.merge(batting,matches,on ='match_no',how='left')
match_data.head()
```

```python
top_runs = batting.groupby('batsman_name')['runs'].sum().sort_values(ascending= False)
print(top_runs)
```

```python
top_runs.head(10).plot(kind='bar')
plt.title("Top 10 Run Scorers")
plt.xlabel("Batsman Name")
plt.ylabel("Runs")
plt.show()
```

```python
top_wickets = bowling.groupby('bowler_name')['wickets'].sum().sort_values(ascending=False)
top_wickets.head(10).plot(kind='bar')
plt.title("Top 10 Wicket Takers")
plt.xlabel("Bowler Name")
plt.ylabel("Wickets")
plt.show()
```

```python
best_economy = bowling.groupby('bowler_name')['economy'].mean().sort_values()
print(best_economy)
```

```python
# Convert batting columns to numeric
batting['runs'] = pd.to_numeric(
    batting['runs'],
    errors='coerce'
)

batting['balls'] = pd.to_numeric(
    batting['balls'],
    errors='coerce'
)

batting['4s'] = pd.to_numeric(
    batting['4s'],
    errors='coerce'
)

batting['6s'] = pd.to_numeric(
    batting['6s'],
    errors='coerce'
)

batting['strike_rate'] = pd.to_numeric(
    batting['strike_rate'],
    errors='coerce'
)
```

```python
# Convert bowling columns
bowling['overs'] = pd.to_numeric(
    bowling['overs'],
    errors='coerce'
)

bowling['maidens'] = pd.to_numeric(
    bowling['maidens'],
    errors='coerce'
)

bowling['runs'] = pd.to_numeric(
    bowling['runs'],
    errors='coerce'
)

bowling['wickets'] = pd.to_numeric(
    bowling['wickets'],
    errors='coerce'
)

bowling['economy'] = pd.to_numeric(
    bowling['economy'],
    errors='coerce'
)
```

```python
strike_rate = batting.groupby(
    'batsman_name'
)['strike_rate'].mean().sort_values(
    ascending=False
)

strike_rate.head(10)
```

```python
team_runs = batting.groupby('team_innings')['runs'].sum().sort_values(ascending=False)
```

```python
team_runs.head(10).plot(kind='bar')
plt.title('Team Wise Runs')
plt.xlabel('Team')
plt.ylabel('Runs')
plt.show()
```

```python
team_wickets = bowling.groupby('bowling_team')['wickets'].sum().sort_values(ascending=False)
team_wickets.head(10).plot(kind='bar')
plt.title('Team Wise Wickets')
plt.xlabel('Team')
plt.ylabel('Wickets')
plt.show()
```
